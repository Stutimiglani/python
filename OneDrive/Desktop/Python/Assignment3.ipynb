```python
# Part A - Dictionaries

inv = {}

# Product 1
n1 = "Mobile Phone"
q1 = 5
p1 = 20000
y1 = 2020

inv["P1"] = {
    "Name": n1,
    "Qty": q1,
    "Price": p1,
    "Year": y1
}

# Product 2
n2 = "Laptop"
q2 = 3
p2 = 50000
y2 = 2022

inv["P2"] = {
    "Name": n2,
    "Qty": q2,
    "Price": p2,
    "Year": y2
}

# Display inventory
print(inv)

# Check release year
print("Year of P1:", inv["P1"]["Year"])
print("Year of P2:", inv["P2"]["Year"])

# Delete release year
del inv["P1"]["Year"]
del inv["P2"]["Year"]

print(inv)
```

```python
# Part B - Tuples

prices = (250, 300, 150, 400, 100, 350, 200)

print("Highest:", max(prices))
print("Lowest:", min(prices))
print("Sum:", sum(prices))

s = sorted(prices)
print("Sorted:", s)

# Tuple cannot be changed
try:
    prices[0] = 500
except TypeError:
    print("Tuple cannot be modified.")
```
