```python
# PROBLEM 4
data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Karan',
             'Neha', 'Arjun', 'Simran', 'Vikas', 'Anjali'],
    'Age': [25, 32, 28, 35, 40, np.nan, 29, 45, 31, 27],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', np.nan, 'Female'],
    'Purchase_Amount': [5000, 7000, 4500, 8000, np.nan,
                        6500, 5500, 9000, 7500, 6000]
}

df = pd.DataFrame(data)

df.to_csv('customer_data.csv', index=False)

df = pd.read_csv('customer_data.csv')

df
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median()
)

df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

print(df)
```

```python
df.isnull().sum()
```

```python
# PROBLEM 5
customers_above_30 = df[df['Age'] > 30]

print(customers_above_30)
```

```python
total_purchase = df.groupby('Gender')['Purchase_Amount'].sum()

print(total_purchase)
```

```python
selected_columns = df[['Customer_ID', 'Name', 'Purchase_Amount']]

print(selected_columns)
```

```python
# PROBLEM 6
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90

print(df)
```

```python
average_purchase = df.groupby('Gender')['Purchase_Amount'].mean()

print(average_purchase)
```

```python
sorted_df = df.sort_values(by='Purchase_Amount', ascending=False)

print(sorted_df)
```
