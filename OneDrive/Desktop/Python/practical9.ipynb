```python
def zipper_file(filename, strings):

    with open(filename, 'w') as file:

        for i in range(0, len(strings), 2):

            zipped_string = strings[i] + strings[i + 1]

            file.write(zipped_string + '\n')


# Example
strings = ['Hello', 'World', 'Python', 'Programming', 'Good', 'Morning']

zipper_file('output.txt', strings)

print("File created successfully!")
```

```python
with open('output.txt', 'r') as file:
    content = file.read()

print(content)
```

```python
def reformat_date(dates):

    months = {
        '01': 'January',
        '02': 'February',
        '03': 'March',
        '04': 'April',
        '05': 'May',
        '06': 'June',
        '07': 'July',
        '08': 'August',
        '09': 'September',
        '10': 'October',
        '11': 'November',
        '12': 'December'
    }

    result = []

    for date in dates:

        month, day, year = date.split('/')

        month_name = months[month]

        formatted_date = "Office Hours on " + month_name + " " + day + ", " + year

        result.append(formatted_date)

    return result
```

```python
dates = [
    '01/15/2025',
    '02/20/2024',
    '03/05/2023',
    '04/25/2022',
    '12/31/2025'
]

result = reformat_date(dates)

for date in result:
    print(date)
```
